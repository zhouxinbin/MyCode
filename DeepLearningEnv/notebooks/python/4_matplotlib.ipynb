```python
x = np.linspace(-1,5,50)
y = x ** 2
plt.plot(x, y, label='y=x**2')
plt.legend(loc='best')

# --- 1. 初始化数据点 ---
# 定义一个x坐标
x0 = 2
# 根据x0计算对应的y坐标（例如，y = 2x + 1）
y0 = 4  # 结果 y0 = 2 * 1 + 1 = 3
# 所以我们关注的点是 (1, 3)

# --- 2. 绘制图表元素 ---

# 创建一个散点图，标记出 (x0, y0) 这个点
# x0, y0: 散点的坐标
# s=50: 散点的大小
# color='b': 散点的颜色为蓝色 ('b' stands for blue)
plt.figure(figsize=(8,6))
plt.scatter(x0, y0, s=50, color='b')

# 绘制一条从 (x0, y0) 到 (x0, 0) 的垂直线
# [x0, x0]: 线的x坐标，这里是 [1, 1]，表示一条垂直线
# [y0, 0]: 线的y坐标，这里是 [3, 0]，表示从y=3到y=0
# 'k--': 线的样式，'k' 表示黑色 (black)，'--' 表示虚线 (dashed line)
# lw=2.5: 线的宽度 (linewidth)
plt.plot([x0, x0], [y0, 0], 'k--', lw=2.5)

# --- 3. 添加文本注解 ---

# # method 1: 使用 plt.annotate()
# plt.annotate() 主要用于指向图表中的特定数据点，并可以添加箭头。
# ###################################
plt.annotate(
    r'$2x+1=%os$' % y0,              # 注解文本内容。
                                     # r'' 表示原始字符串，用于处理LaTeX语法中的反斜杠。
                                     # $...$ 启用LaTeX数学模式，使文本渲染为公式。
                                     # %os 会将y0的值（这里是3）格式化插入到字符串中。
                                     # 最终显示为 "2x+1=3" (公式样式)。
    xy=(x0, y0),                     # 注解箭头指向的数据点坐标 (目标点)。这里是 (1, 3)。
    xycoords='data',                 # 指定 xy 坐标系为数据坐标系（默认值，但明确写出）。
    xytext=(+30, -30),               # 注解文本自身的位置。这是相对于 xy 点的偏移量。
                                     # (+30, -30) 表示从 xy 点向右偏移30像素，向下偏移30像素。
    textcoords='offset points',      # 指定 xytext 的单位是像素点数，相对于 xy。
    fontsize=16,                     # 注解文本的字体大小。
    arrowprops=dict(                 # 字典用于定义箭头的属性。
        arrowstyle='->',             # 箭头样式，'->' 表示从文本指向目标点。
        connectionstyle='arc3,rad=.2' # 箭头连接线的样式。'arc3' 表示一个三阶贝塞尔曲线，
                                     # 'rad=.2' 控制曲线的弯曲程度（弧度）。
    )
)

# # method 2: 使用 plt.text()
# plt.text() 主要用于在图表上任意指定位置放置文本，通常不带箭头。
# ###################################
plt.text(
    -3.7, 3,                         # 文本放置的绝对坐标 (x, y)。
    r'$This\ is\ the\ some\ text.\ \mu\ \sigma_i\ \alpha_t$', # 文本内容。
                                     # 同样使用 r'' 和 $...$ 来处理LaTeX数学公式。
                                     # \ 是LaTeX中表示一个空格的命令。
                                     # \mu, \sigma_i, \alpha_t 是LaTeX命令，用于显示希腊字母和带下标的符号。
    fontdict={'size': 16, 'color': 'r'} # 字典用于定义文本的字体属性。
                                     # 'size': 字体大小。
                                     # 'color': 字体颜色，'r' 表示红色 (red)。
)

# --- 4. 显示图表 ---
# 显示所有绘制的元素，包括散点、虚线和两种注解文本。
plt.show()
```

```python
# bbox

# --- 1. 数据准备 ---

# 使用 NumPy 生成数据
# np.linspace(start, stop, num) 在指定区间内生成均匀分布的 num 个点。
# 这里从 -3 到 3 生成了 50 个点作为 x 坐标。
x = np.linspace(-3, 3, 50)
# 根据 x 坐标计算 y 坐标，这是一个简单的线性关系：y = 0.1x
y = 0.1 * x

# --- 2. 绘制图表及初始设置 ---

# 创建一个新的图表窗口（Figure），如果没有指定，Matplotlib 会自动创建一个。
plt.figure()
# 绘制线图
# plt.plot(x, y): 使用 x 和 y 数据绘制一条线。
# linewidth=10: 设置线的宽度为 10，使线非常粗。
plt.plot(x, y, linewidth=10)
# 设置 y 轴的显示范围
# plt.ylim(ymin, ymax): 将 y 轴的范围设置为从 -2 到 2。
plt.ylim(-2, 2)

# --- 3. 坐标轴（Spines）和刻度（Ticks）的定制 ---

# 获取当前图表的 Axes 对象。Axes 是实际绘图区域。
# 所有的轴、刻度、标签等都属于这个 Axes 对象。
ax = plt.gca()

# 隐藏右侧和顶部的坐标轴线（Spines）
# Matplotlib 的图表默认有四条边框线，称为 'spines'：'left', 'right', 'bottom', 'top'。
# ax.spines['right'].set_color('none'): 将右侧边框线的颜色设置为 'none'，使其不可见。
ax.spines['right'].set_color('none')
# ax.spines['top'].set_color('none'): 将顶部边框线的颜色设置为 'none'，使其不可见。
# 注意：代码中重复了这一行，是冗余的，但不会引起错误。
ax.spines['top'].set_color('none')
ax.spines['top'].set_color('none') # 这一行是重复的，效果同上一行

# 设置 x 轴刻度标签的位置
# ax.xaxis.set_ticks_position('bottom'): 仅在底部显示 x 轴的刻度标记和标签。
# 默认情况下，如果顶部也有边框，刻度可能会在顶部和底部都显示。
ax.xaxis.set_ticks_position('bottom')
# 移动底部坐标轴线的位置到 y=0 的数据坐标处
# ax.spines['bottom'].set_position(('data', 0)): 将底部边框线的位置设置为数据坐标系中的 y=0。
# ('data', 0) 表示基于数据值来定位，y=0。这使得 x 轴（底部边框）穿过原点。
ax.spines['bottom'].set_position(('data', 0))

# 设置 y 轴刻度标签的位置
# ax.yaxis.set_ticks_position('left'): 仅在左侧显示 y 轴的刻度标记和标签。
ax.yaxis.set_ticks_position('left')
# 移动左侧坐标轴线的位置到 x=0 的数据坐标处
# ax.spines['left'].set_position(('data', 0)): 将左侧边框线的位置设置为数据坐标系中的 x=0。
# ('data', 0) 表示基于数据值来定位，x=0。这使得 y 轴（左侧边框）穿过原点。
ax.spines['left'].set_position(('data', 0))

# --- 4. 刻度标签（Tick Labels）的样式定制 ---

# 遍历所有 x 轴和 y 轴的刻度标签
# ax.get_xticklabels(): 获取所有 x 轴的刻度标签对象（Text 对象列表）。
# ax.get_yticklabels(): 获取所有 y 轴的刻度标签对象（Text 对象列表）。
# + 运算符在这里用于连接两个列表，形成一个包含所有刻度标签的大列表。
for label in ax.get_xticklabels() + ax.get_yticklabels():
    # 设置刻度标签的字体大小
    # label.set_fontsize(12): 将当前刻度标签的字体大小设置为 12。
    label.set_fontsize(12)
    # 为刻度标签设置背景框（bounding box）
    # label.set_bbox(dict(...)): 接受一个字典，定义背景框的属性。
    # facecolor='white': 背景框的填充颜色为白色。
    # edgecolor='None': 背景框的边框颜色设置为 'None'，即没有可见边框。
    # alpha=0.7: 背景框的透明度设置为 0.7（70%不透明，30%透明），
    #           这样可以略微看到背景框后面的图表内容，防止完全遮挡。
    label.set_bbox(dict(facecolor='white', edgecolor='None', alpha=0.7))

# --- 5. 显示图表 ---

# 显示所有绘制的元素，包括线图和定制后的坐标轴、刻度标签。
plt.show()
```

```python
# 柱状图

# --- 1. 数据准备 ---
n = 12 # 定义数据点的数量，将用于生成 x 轴的范围和 y 值
X = np.arange(n) # 生成一个从 0 到 n-1 的整数数组，作为柱状图的 x 坐标

# 计算第一组 Y 值 (Y1)
# (1 - X / float(n)): 创建一个从接近 1 递减到接近 0 的序列，赋予“趋势”
# np.random.uniform(0.5, 1.0, n): 生成 n 个在 0.5 到 1.0 之间均匀分布的随机数，作为“波动”
# Y1 是趋势和波动的乘积，使其具有递减的随机值
Y1 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)

# 计算第二组 Y 值 (Y2)，与 Y1 类似，但使用不同的随机数
Y2 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)

# --- 2. 绘制柱状图 ---

# 绘制第一组柱状图 (向上)
# X: 柱子的 x 坐标
# +Y1: 柱子的高度 (正值，向上生长)
# facecolor='#9999ff': 柱子的填充颜色 (淡蓝色/紫色)
# edgecolor='white': 柱子的边框颜色 (白色)
plt.bar(X, +Y1, facecolor='#9999ff', edgecolor='white')

# 绘制第二组柱状图 (向下)
# X: 柱子的 x 坐标
# -Y2: 柱子的高度 (负值，向下生长)
# facecolor='#ff9999': 柱子的填充颜色 (淡粉色/红色)
# edgecolor='white': 柱子的边框颜色 (白色)
plt.bar(X, -Y2, facecolor='#ff9999', edgecolor='white')

# --- 3. 为柱状图添加数值标签 ---

# 遍历第一组数据 (X, Y1)，为每个柱子添加文本标签
for x, y in zip(X, Y1):
    # ha: horizontal alignment (水平对齐方式)
    # plt.text(x_pos, y_pos, text_string, **kwargs) 在指定位置添加文本
    # x + 0.4: 文本的 x 坐标，相对于柱子中心稍微向右偏移，使其在柱子上方水平居中
    # y + 0.05: 文本的 y 坐标，在柱子顶端之上再加 0.05 的偏移量
    # '%.2f' % y: 格式化 y 值，显示为浮点数，保留两位小数
    # ha='center': 文本内容相对于 x_pos 水平居中
    # va='bottom': 文本内容的底部与 y_pos 对齐，这样文本就会在 y_pos 之上
    plt.text(x + 0.04, y + 0.05, '%.2f' % y, ha='center', va='bottom')

# 遍历第二组数据 (X, Y2)，为每个柱子添加文本标签
for x, y in zip(X, Y2):
    # ha: horizontal alignment (水平对齐方式)
    # plt.text(x_pos, y_pos, text_string, **kwargs)
    # x + 0.4: 文本的 x 坐标，与上方标签保持一致的水平偏移
    # -y - 0.05: 文本的 y 坐标。注意这里用的是 -y (因为柱子是向下的)，再减去 0.05 的偏移量，
    #            使其在柱子底端之下。
    # '%.2f' % y: 格式化原始正值 y，显示为浮点数，保留两位小数 (通常显示绝对值)
    # ha='center': 文本内容相对于 x_pos 水平居中
    # va='top': 文本内容的顶部与 y_pos 对齐，这样文本就会在 y_pos 之下
    plt.text(x + 0.04, -y - 0.05, '-%.2f' % y, ha='center', va='top')

# --- 4. 设置图表轴的显示范围和刻度 ---

# 设置 x 轴的显示范围
# 从 -0.5 开始，到 n (12) 结束，为柱子两边留出一些空白
plt.xlim(-.5, n)
# 移除 x 轴的所有刻度标记和标签，使图表更简洁
plt.xticks([])

# 设置 y 轴的显示范围
# 从 -1.25 到 1.25，为上下两组柱子及其标签留出足够的空间
plt.ylim(-1.25, 1.25)
# 移除 y 轴的所有刻度标记和标签
plt.yticks([])

# --- 5. 显示图表 ---
plt.show() # 显示绘制好的图表
```

```python
# 等高线图

# 定义高度函数
def f(x,y):
    return (1 - x / 2 + x**5 + y**3) * np.exp(-x**2 -y**2)

n = 256  # 设置网格密度
x = np.linspace(-3, 3, n)  # 生成 x 轴坐标点范围
y = np.linspace(-3, 3, n)  # 生成 y 轴坐标点范围
X, Y = np.meshgrid(x, y)  # 生成网格点坐标矩阵

# 绘制填充的等高线图
# X, Y 为坐标, f(X,Y) 为高度值, 8 为等高线层数
# alpha=0.75 设置透明度, cmap=plt.cm.hot 设置热力图配色
plt.contourf(X, Y, f(X, Y), 8, alpha=0.75, cmap=plt.cm.hot)

# 绘制等高线线条
# 使用黑色线条，线宽为 0.5
C = plt.contour(X, Y, f(X, Y), 8, colors='black', linewidths=.5)

# 为等高线添加数值标签
# inline=False 表示标签不会打断线条，fontsize=10 设置字体大小
plt.clabel(C, inline=False, fontsize=10)

plt.xticks(())  # 隐藏 x 轴刻度
plt.yticks(())  # 隐藏 y 轴刻度
plt.show()      # 显示图表
```

```python
# 3D
from mpl_toolkits.mplot3d import Axes3D

# 添加坐标轴系统
fig = plt.figure()
ax = Axes3D(fig)

# X Y Value
X = np.arange(-4, 4, 0.25)
Y = np.arange(-4, 4, 0.25)
# 二维矩阵：X横坐标 Y纵坐标
X, Y = np.meshgrid(X, Y)

# Z轴数据
R = np.sqrt(X ** 2 + Y ** 2)
Z = np.sin(R)

# 绘制 3D
ax.plot_surface(X, Y, Z, 
                rstride = 1, # 
               cstride = 1, # 全部采样
               cmap=plt.get_cmap('rainbow')) # 颜色映射表
# 设置等高线图
ax.contourf(X, Y, Z, 
            zdir='z', # 投影 z 轴
            offset = -2, #偏移 z=0 的位置
            cmap='rainbow')
ax.set_zlim(-2, 2) # 设置z范围

plt.show()
```

```python
# 主次坐标轴


# --- 1. 数据准备 ---
# 创建 X 轴数据：从 0 到 10，步长为 0.1 的等差数列
x = np.arange(0, 10, 0.1)
# 创建第一组 Y 轴数据 (y1)：一个开口向上的抛物线
y1 = 0.05 * x**2
# 创建第二组 Y 轴数据 (y2)：y1 的相反数，一个开口向下的抛物线
y2 = -1 * y1

# --- 2. 创建画布和主坐标轴 ---
# plt.subplots() 是创建一个新图形的标准方式。
# fig 是整个画布对象。
# ax1 是画布上的第一个坐标系对象（主坐标轴），默认其 Y 轴在左侧。
fig, ax1 = plt.subplots()

# --- 3. 关键步骤：创建次坐标轴 ---
# 【核心代码】
# ax1.twinx() 的作用是产生一个新的坐标系对象 (ax2)。
# 这个新坐标系 ax2 与原坐标系 ax1 共享同一个 X 轴（即 X 轴刻度完全同步）。
# 但是，ax2 会拥有一个独立的 Y 轴，并且这个 Y 轴默认会被放置在图表的右侧。
ax2 = ax1.twinx()

# --- 4. 在不同的坐标轴上绘图 ---
# 在主坐标轴 (ax1, 左侧 Y 轴) 上绘制 y1 数据。
# 'g-' 表示用绿色的实线绘制。
ax1.plot(x, y1, 'g-')

# 在次坐标轴 (ax2, 右侧 Y 轴) 上绘制 y2 数据。
# 'b--' 表示用蓝色的虚线绘制。
# 注意：这里使用的是 ax2 对象来调用 plot 方法。
ax2.plot(x, y2, 'b--')

# --- 5. 设置标签和样式 ---
# 设置共享的 X 轴标签（只需设置一次，通常在 ax1 上设置）
ax1.set_xlabel('X data')

# 设置主 Y 轴 (左侧) 的标签。
# color='g' 将标签文字设置为绿色，以便与绿色的曲线对应，增加可读性。
ax1.set_ylabel('Y1', color='g')

# 设置次 Y 轴 (右侧) 的标签。
# color='b' 将标签文字设置为蓝色，以便与蓝色的曲线对应。
ax2.set_ylabel('Y2', color='b')

# --- 6. 显示图表 ---
plt.show()
```
