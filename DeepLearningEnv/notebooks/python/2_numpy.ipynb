```python
import numpy as np

array = np.array([[1,2,3], [4,5,6]])

print("array =\n", array)
print("array dimension =", array.ndim)
print("array shape =", array.shape)
print("array item size =", array.size)
```

```python
# array create

array = np.array([1,2,3], dtype=int)
print(array, array.dtype)

array = np.array([[1,2,3],[4,5,6]])
print(array)

array = np.zeros((2,3))
print(array)

array = np.ones((2,3))
print(array)

array = np.arange(12).reshape((3,4))
print(array)

# 线段 0 ~ 10 分成 11 个点，10 个线段
array = np.linspace(0, 10, 11)
# 打印的是点
print(array)
```

```python
# numpy 基础运算

a = np.array([2,3,4,5])
b = np.arange(4)
# 基础运算
c = a * b
print(c)
# 三角函数
d = np.cos(a)
print(d)
# 判断数组内元素大小情况
print(d > 0)
# 矩阵点乘
aa = a.reshape((2,2))
bb = b.reshape((2,2))
cc = np.dot(aa, bb)
print(cc)
# 随机矩阵
rdm = np.random.random((2,4))
print(rdm)
print(np.sum(rdm)) # 默认全部元素
print(np.max(rdm))
print(np.min(rdm))

print(np.sum(rdm, axis=0)) # 按列分别求和
print(np.sum(rdm, axis=1)) # 按行分别求和
```

```python
# 元素索引
A = np.arange(2,14).reshape((3,4))
print(np.argmax(A))
print(np.argmin(A))
# 中位数
print(np.median(A))
# 平均值
print(np.average(A))
# 前缀和
print(np.cumsum(A))
# 差分
print(np.diff(A))
# 非零数位置
print(np.nonzero(A))
```

```python
# 转置
A = np.arange(14,2,-1).reshape((3,4))
print(A)
print(np.transpose((A)))

# 排序
print(np.sort(A))

# 限制大小
print(np.clip(A,5,9))

# 按行列计算平均值
print(np.mean(A,axis=1))
```

```python
# 索引
B = np.arange(2,14).reshape((3,4))
print(B)
print(B[1][2])
print(B[1,2])

# 遍历行
for row in B:
    print(row)

# 遍历列
for col in B.T:
    print(col)

# 遍历元素
for item in B.flat:
    print(item)
```
