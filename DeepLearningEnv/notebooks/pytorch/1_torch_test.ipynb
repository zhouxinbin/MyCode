pytorch 官方文档: https://docs.pytorch.org/docs/stable/index.html

```python
import torch

# 打印 PyTorch 版本
print("PyTorch Version:", torch.__version__)

# 检查 CUDA (GPU支持) 是否可用
# 对于我们的 CPU 环境，这里的结果应该是 False，这才是正确的！
cuda_available = torch.cuda.is_available()
print(f"CUDA Available: {cuda_available}")

# 创建一个简单的张量 (tensor) 来验证核心功能
x = torch.rand(5, 3)
print("\nSuccessfully created a random tensor:")
print(x)
```
