```python
import torch
import torchvision
from torch import nn
from torch.nn import Conv2d
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
from torch.nn import MaxPool2d
```

```python
# 1. 定义原始数据：模拟一个 5x5 的输入图像
input = torch.tensor([[1, 2, 0, 3, 1], 
                      [0, 1, 2, 3, 1], 
                      [1, 2, 1, 0, 0], 
                      [5, 2, 3, 1, 1], 
                      [2, 1, 0, 1, 1]], dtype=torch.float32)

input = torch.reshape(input, (-1, 1, 5, 5))
# print(input.shape)

class Poll(nn.Module):
    def __init__(self):
        super(Poll, self).__init__()
        self.max_pool_2d = MaxPool2d(kernel_size=3, ceil_mode=True)

    def forward(self, input):
        output = self.max_pool_2d(input)
        return output

poll = Poll()
output = poll(input)
print(output)
```

```python
class Pool(nn.Module):
    def __init__(self):
        super(Pool, self).__init__()
        self.max_pool_2d = MaxPool2d(kernel_size=3, ceil_mode=True)

    def forward(self, input):
        output = self.max_pool_2d(input)
        return output

dataset = torchvision.datasets.CIFAR10(root="/workspace/data", train=False, transform=torchvision.transforms.ToTensor(), download=True)
dataloader = DataLoader(dataset, batch_size=64)

writer = SummaryWriter("logs")
step = 0

pool = Pool()

for data in dataloader:
    img, target = data
    writer.add_images("input", img, step)
    output = pool(img)
    writer.add_images("output", output, step)
    step = step + 1
writer.close()
```
