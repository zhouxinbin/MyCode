````python
# 展示 image
import numpy as np
from PIL import Image

# 第一个蚂蚁
writer = SummaryWriter("logs")

ant1_path = "/workspace/data/hymenoptera_data/train/ants/0013035.jpg"
ant1_PIL = Image.open(ant1_path)
ant1_array = np.array(ant1_PIL)

print(type(ant1_array))
print(ant1_array.shape)

writer.add_image("ants", ant1_array, 1, dataformats="HWC")

# 第二个 蚂蚁
ant2_path = "/workspace/data/hymenoptera_data/train/ants/5650366_e22b7e1065.jpg"
ant2_PIL = Image.open(ant2_path)
ant2_array = np.array(ant2_PIL)

print(type(ant2_array))
print(ant2_array.shape)

writer.add_image("ants", ant2_array, 2, dataformats="HWC")

# 第一个蜜蜂
bees1_path = "/workspace/data/hymenoptera_data/train/bees/16838648_415acd9e3f.jpg"
bees1_PIL = Image.open(bees1_path)
bees1_array = np.array(bees1_PIL)

print(type(bees1_array))
print(bees1_array.shape)

writer.add_image("bees", bees1_array, 1, dataformats="HWC")
````
