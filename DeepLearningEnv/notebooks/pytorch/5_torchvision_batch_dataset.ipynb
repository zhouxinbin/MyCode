```python
# 数据集转换为 tensor
dataset_transform = torchvision.transforms.Compose([torchvision.transforms.ToTensor()])

train_set = torchvision.datasets.CIFAR10(root="/workspace/data", train="True", 
                                         transform=dataset_transform, download="True")
test_set = torchvision.datasets.CIFAR10(root="/workspace/data", train="False", 
                                        transform=dataset_transform, download="True")

print(test_set[0])
```

```python
# 使用 TensorBoard 显示数据集
from torch.utils.tensorboard import SummaryWriter
writer = SummaryWriter("logs")

for i in range(10):
    img, target = test_set[i]
    writer.add_image("CIFAR-10", img, i)

writer.close()
```
