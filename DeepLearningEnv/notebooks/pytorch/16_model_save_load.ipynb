```python
import torchvision
import torch

vgg16 = torchvision.models.vgg16(pretrained = False)
# 方式一：模型结构 + 模型参数
torch.save(vgg16, "models/vgg16_save_1.pth")
# 方式二：模型参数
torch.save(vgg16.state_dict(), "models/vgg16_save_2.pth")
```

```python
import torch
import torchvision

# 加载模型 1
model_1 = torch.load("models/vgg16_save_1.pth", weights_only=False)
print(model_1)

# 加载模型 2
vgg16 = torchvision.models.vgg16(pretrained="False")
vgg16.load_state_dict(torch.load("models/vgg16_save_2.pth", weights_only=False))
print(vgg16)
```
