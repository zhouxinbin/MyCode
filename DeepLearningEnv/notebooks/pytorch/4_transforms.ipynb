```python
from torchvision import transforms
from PIL import Image
from torch.utils.tensorboard import SummaryWriter
```

```python
img_path = "/workspace/data/hymenoptera_data/train/ants/0013035.jpg"
img = Image.open(img_path)
print(img)
```

## ToTensor

```python
writer = SummaryWriter("logs")

tensor_trans = transforms.ToTensor()
img_tensor = tensor_trans(img)
# print(img_tensor)

writer.add_image("Tensor_img", img_tensor)
writer.close()
```

## Normalize

```python
writer = SummaryWriter("logs")

print(img_tensor[0][0][0])
norm_trans = transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])
img_norm = norm_trans(img_tensor)
print(img_norm[0][0][0])

writer.add_image("Normalize", img_norm)
writer.close()
```

## Resize

```python
print(img.size)

trans_resize = transforms.Resize((512, 512))
img_resize = trans_resize(img)

print(img_resize.size)

writer.add_image("Resize", tensor_trans(img_resize))
```

## Compose
流水线操作

```python
# 定义缩放，短边缩放到 512 其他自适应
trans_resize_2 = transforms.Resize(512)
# compose 流水线函数：缩放 并且 转换为 tensor 格式
trans_compose = transforms.Compose([trans_resize_2, tensor_trans])
# 执行流水线函数
img_resize_2 = trans_compose(img)
# 上传结果到 TensorBoard
writer.add_image("Compose", img_resize_2, 1)
```

## RandomCrop
随机裁剪

```python
random_crop = transforms.RandomCrop((300, 300))
trans_compose_2 = transforms.Compose([random_crop, tensor_trans])
for i in range(10):
    img_crop = trans_compose_2(img)
    writer.add_image("RandomCrop", img_crop, i)
```
